## 2. Variables

A **variable** is a named box that holds a value. In Java, every variable has a **type** that says what kind of value it can hold.

| Type | Holds | Example |
|---|---|---|
| `int` | whole numbers | `int credits = 5;` |
| `double` | numbers with decimals | `double price = 4.25;` |
| `String` | text | `String course = "CS 142";` |
| `boolean` | `true` or `false` | `boolean isOpen = true;` |
| `char` | a single character | `char grade = 'A';` |

The pattern is always: **type name = value;**

````java
int credits = 5;
double price = 4.25;
String course = "CS 142";
boolean isOpen = true;
char grade = 'A';

System.out.println(credits);
System.out.println(price);
System.out.println(course);
System.out.println(isOpen);
System.out.println(grade);
````

The notebook remembers variables between cells. This cell uses `course` and `credits` from the cell above.

The `+` sign **joins** (concatenates) text and values into one String:

````java
System.out.println("I am taking " + course + " for " + credits + " credits.");
````

Variables can **change**. Assign a new value with `=` (you don't repeat the type):

````java
credits = 10;
System.out.println("Now credits is " + credits);
````

## 3. Doing math with variables

Java does arithmetic with `+  -  *  /` and `%` (remainder).

````java
double mealCost = 40.0;
int tipPercent = 20;

double tip = mealCost * tipPercent / 100.0;
double total = mealCost + tip;

System.out.println("Tip: " + tip);
System.out.println("Total: " + total);
````

### ⚠️ Watch out: integer division
When you divide two `int`s, Java **drops the decimal part**. Run this and look closely at the results:

````java
System.out.println(7 / 2);      // two ints
System.out.println(7 / 2.0);    // one double
System.out.println(7 % 2);      // remainder
````

## 5. From notebook cells to a real Java program

So far you've typed statements straight into cells. A real `.java` file can't do that. Java needs every statement to live **inside a method, inside a class**.

Here is the smallest complete Java program. It has the same shape as `HelloWorld.java` in Part 2:

```java
public class Greeting {                         // (1) the class
    public static void main(String[] args) {    // (2) the main method
        System.out.println("Hi there!");        // (3) your code goes here
    }                                           // closes main
}                                               // closes the class
```

1. **The class.** `public class Greeting` wraps the whole program. The class name must match the file name, so this program would live in `Greeting.java`.
2. **The main method.** `public static void main(String[] args)` is where the program **starts**. When you click **Run**, Java finds this line and runs the statements inside its braces, top to bottom. For now, copy this line exactly. You'll learn what `static`, `void` and `String[] args` mean later in the course.
3. **Your code.** Everything you practiced in sections 1–3 goes **inside main's braces**, between `{` and `}`.

Braces always come in pairs. The inner pair belongs to `main`, and the outer pair belongs to the class. Indenting each level makes it easy to see which is which.

A notebook can hold a whole program too. Run the cell below. Notice that it prints **nothing**: it only *defines* the program.

````java
public class Greeting {
    public static void main(String[] args) {
        System.out.println("Hi there!");
        System.out.println("This line runs second.");
    }
}
````

To start the program in a notebook, you call its `main` method yourself. In Part 2 the **Run** button does this for you. (`new String[0]` just means "no extra input." You can ignore it for now.)

````java
Greeting.main(new String[0]);
````

### Variables inside main

Variables, math and `+` work exactly the same inside `main`. Here is a small program that follows the same pattern as `Variables.java`: **declare variables, calculate, then print**. Run it and read it line by line.

````java
public class CoffeeOrder {
    public static void main(String[] args) {
        // 1. Declare variables
        String drink = "Latte";
        int cups = 3;
        double pricePerCup = 4.5;

        // 2. Calculate with them (no typing the answer!)
        double subtotal = cups * pricePerCup;

        // 3. Print, joining labels and variables with +
        System.out.println("Drink: " + drink);
        System.out.println("Cups: " + cups);
        System.out.println("Subtotal: " + subtotal);
    }
}

CoffeeOrder.main(new String[0]);
````

One difference from the notebook: variables declared inside `main` exist only inside `main`. Each time the program runs, it starts fresh, so a program must declare every variable it uses.
