## 2. Variables

A **variable** is a named box that holds a value. In Java, every variable has a **type** that says what kind of value it can hold.

| Type | Holds | Example |
|---|---|---|
| `int` | whole numbers | `int credits = 5;` |
| `double` | numbers with decimals | `double price = 4.25;` |
| `String` | text | `String course = "CS 142";` |
| `boolean` | `true` or `false` | `boolean isOpen = true;` |
| `char` | a single character | `char grade = 'A';` |

The pattern is always: **type name = value;**

````java
int credits = 5;
double price = 4.25;
String course = "CS 142";
boolean isOpen = true;
char grade = 'A';

System.out.println(credits);
System.out.println(price);
System.out.println(course);
System.out.println(isOpen);
System.out.println(grade);
````

The notebook remembers variables between cells. This cell uses `course` and `credits` from the cell above.

The `+` sign **joins** (concatenates) text and values into one String:

````java
System.out.println("I am taking " + course + " for " + credits + " credits.");
````

Variables can **change**. Assign a new value with `=` (you don't repeat the type):

````java
credits = 10;
System.out.println("Now credits is " + credits);
````

## 3. Doing math with variables

Java does arithmetic with `+  -  *  /` and `%` (remainder).

````java
double mealCost = 40.0;
int tipPercent = 20;

double tip = mealCost * tipPercent / 100.0;
double total = mealCost + tip;

System.out.println("Tip: " + tip);
System.out.println("Total: " + total);
````

### ⚠️ Watch out: integer division
When you divide two `int`s, Java **drops the decimal part**. Run this and look closely at the results:

````java
System.out.println(7 / 2);      // two ints
System.out.println(7 / 2.0);    // one double
System.out.println(7 % 2);      // remainder
````

**Think about it:** Why did `7 / 2` give `3` and not `3.5`? What would `tipPercent / 100` give if `tipPercent` is `20`? Try it in the cell below to check your guess.

````java
System.out.println(tipPercent / 100);
````
